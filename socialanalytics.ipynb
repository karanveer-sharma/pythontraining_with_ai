```python
import numpy as np
import pandas as pd
```

```python
social = {
    "Handle": ["@techguru", "Gamer_123", "@Beauty_lush","Travel_Boy"],
    "Followers": ["1.2M","500k","2.5M","900"],
    "Avg_likes":["50k","12000", "150k","45" ],
    "Category":["Tech","gaming","Beauty","travel"],
    "Verified":["Yes","No","Yes","No"]
    }
```

```python
df_social=pd.DataFrame(social)
print("social Media data")
display(df_social)
```

```python
df_social["Handle"]=df_social["Handle"].str.replace("@","", regex=False)
```

```python
df_social["Handle"].head()
```

```python
def convert_to_no(value_str):
  value_str = str(value_str).lower()
  if "m"in value_str:
    return float(value_str.replace("m",""))*1000000
  elif"k" in value_str:
    return float(value_str.replace("k",""))*1000
  else:
    return float(value_str)
```

```python
df_social["Followers_num"]=df_social["Followers"].apply(convert_to_no)
df_social["Avg_likes_num"]=df_social["Avg_likes"].apply(convert_to_no)
df_social.head()
```

```python
df_social["Category"]=df_social["Category"].str.capitalize()
df_social["Category"].head()
```

```python
df_social["Followers_num"].head()
```

```python
df_social["Engagement_rate"]= (df_social["Avg_likes_num"]/df_social["Followers_num"])*100
top_creators= df_social[df_social["Engagement_rate"]>3.0]
print("\ntop creators")
display(top_creators)
```

```python
Log_data={
    "Log_entry": [
        "192.168.1.1 - GET /login - 200 OK - 15ms",
        "192.168.1.5 - POST /upload - 500 Error - 2000ms",
        "10.0.0.4 - GET /home - 200 OK - 12ms",
        "192.168.1.15 - GET/admin - 403 Forbidden - 5ms",
        "10.0.0.8 -GET/dashboard - 200 OK - 1100ms"    ]
}
```

```python
df_log=pd.DataFrame(Log_data)
print("log data")
display(df_log)
```

```python
parsed_logs=df_log["Log_entry"].str.split(" - ", expand=True)
parsed_logs.columns =["Ip_address","Request","Status","Latency"]
print("\nParsed logs")
display(parsed_logs)
```

```python
parsed_logs["Latency_value"] = parsed_logs["Latency"].astype(str).str.replace("ms", "").astype(float)
parsed_logs["Latency_value"].head()
```

```python
suspicious= parsed_logs[parsed_logs["Status"].str.contains("403")]
print("\n security alert (admin access ) :")
display(suspicious[["Ip_address", "Request"]])
```

```python
slow_query =parsed_logs[parsed_logs["Latency_value"]>1000]
print("\n slow response")
display(slow_query[["Request","Latency_value"]])
```
