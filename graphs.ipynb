```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
fig= plt.figure(figsize=(2,5))
ax = plt.axes()
plt.show()
```

```python
x = np.linspace(0,10,20)
y= x**2
plt.plot(x,y,label="Growth", color="green",marker="o",ls=":",lw=4)
plt.title("Revenue Trend")
plt.legend()
plt.show()
```

```python
courses=["Ai","Data analytic", "Web dev"]
students=[45,30,25]
plt.bar(courses, students,color=["red","blue","green"])
plt.title("Students enrolled in courses")
plt.xlabel("COURSES")
plt.ylabel("STUDENTS")
plt.xticks(rotation=45)
plt.show()
```

```python
data={
    "Category":["A","B","C","D"],
    "Value1":[10,12,8,15],
    "Value2":[5,8,10,7],
    "Value3":[3,4,6,5]
}
df = pd.DataFrame(data)
df= df.set_index("Category")
df.plot(kind="bar",stacked=True, ax=plt.gca())
plt.title("Stacked Bar Chart")
plt.show()
```

```python
hours=np.random.randint(1,10,50)
scores=hours*10+np.random.randint(0,5,50)
plt.scatter(hours,scores,label="hours vs scores",alpha=1,color="blue",edgecolor="black")
plt.title("Hours vs Scores")
plt.xlabel("Hours")
plt.ylabel("Scores")
plt.legend()
plt.show()
```

```python
ages=np.random.normal(22,3,1000)
plt.hist(ages,bins=10,color="purple",edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
scores=[12,22,23,44,45,56,66,77,88,90,55,45,120,321]
plt.figure(figsize=(10,6))
plt.boxplot(scores,vert=False,patch_artist=True,labels=["Scores"],boxprops=dict(facecolor="blue",color="black"))
plt.title("Box Plot")
plt.show()
```

```python
ai_student = np.random.normal(30,5,100)
web_student = np.random.normal(25,4,100)
ds_students= np.concatenate([np.random.normal(10,2,50),np.random.normal(30,2,50)])
data=[ai_student,web_student,ds_students]
labels=["AI","Web","DS"]
plt.figure(figsize=(10,6))
parts=plt.violinplot(data,showmeans=False,showmedians=True,showextrema=True,vert=True)
colors=["blue","red","green"]
for i,pc in enumerate(parts["bodies"]):
    pc.set_facecolor(colors[i])
    pc.set_edgecolor("black")
    pc.set_alpha(0.7)
plt.xticks([1,2,3],labels)
plt.title("Violin Plot",fontsize=14)
plt.xlabel("Students")
plt.ylabel("Score")
plt.grid(axis="y",linestyle="--",alpha=0.5)
plt.show()
```
