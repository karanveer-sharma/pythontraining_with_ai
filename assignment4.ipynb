```python
import pandas as pd
import numpy as np

# Dataset 1: Employees Table
emp_data = {
    'Emp_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Rajan', 'Anurag', 'Manya Gupta', 'Priya', 'Abhinav Garg', 'Frank'],
    'Dept_ID': ['D1', 'D1', 'D2', 'D3', 'D2', np.nan], # Frank has no Dept
    'Salary': [70000, 80000, 120000, 90000, 115000, 60000],
    'Join_Date': ['2023-01-15', '2023-02-10', '2022-05-20', '2021-08-15', '2024-01-01', '2023-06-30']
}
df_emp = pd.DataFrame(emp_data)

# Dataset 2: Departments Table
dept_data = {
    'Dept_ID': ['D1', 'D2', 'D3', 'D4'],
    'Dept_Name': ['HR', 'Engineering', 'Marketing', 'Sales'],
    'Location': ['Banglore', 'Hyderabad', 'Noida', 'Gurugram']
}
df_dept = pd.DataFrame(dept_data)

print("--- Employees ---")
display(df_emp)
print("\n--- Departments ---")
display(df_dept)
```

```python
new_hire_data = {
    'Emp_ID': [107],
    'Name': ['Harman'],
    'Dept_ID': ['D4'],
    'Salary': [90000],
    'Join_Date': ['2025-09-10']
}
new_hire = pd.DataFrame(new_hire_data)
df_emp = pd.concat([df_emp, new_hire], ignore_index=True)
display(df_emp)
```

```python
df_full = pd.merge(df_emp, df_dept, on='Dept_ID', how='left')
display(df_full)
```

```python
mean_salary=df_full.groupby(df_full["Dept_ID"])["Salary"].mean()
display(mean_salary)
```

```python
df_full["Avg_salary"]=df_full["Dept_ID"].map(mean_salary)
display(df_full)
```

```python
higher_salary=df_full[df_full["Salary"]>df_full["Avg_salary"]]
print("Employees with higher salary than department average:")
display(higher_salary)
```

```python
dp_loc=df_full.groupby(df_full["Dept_ID"])["Location"].first()
display(dp_loc)
```

```python
bng=df_full.loc[df_full["Location"]=="Banglore"]
display(bng)
```

```python
df_pivot=pd.pivot_table(df_full,values="Salary",index="Dept_ID",columns="Location",aggfunc="mean")
display(df_pivot)
```

```python
df_melt=pd.melt(df_pivot.reset_index(),id_vars="Dept_ID",var_name="Location",value_name="Avg_Salary")
display(df_melt)
```

```python
print(df_full["Dept_Name"].memory_usage(deep=True),"bytes")
```

```python
df_full["Bonus_vec"]=df_full["Salary"]*0.10
df_full["Bonus_vec"].head(3)
```

```python
def complex_bonus(x):
  if x>100000:
    return x*0.15
  else:
    return x*0.05
df_full["Bonus_Complex"]= df_full["Salary"].apply(complex_bonus)
df_full["Bonus_Complex"].head(3)
```

```python
df_full["First_name"]=df_full["Name"].str.split(' ').str[0]
hr_employes=df_full[df_full["Dept_Name"].str.contains("Eng", na = False)]
print("Engineering")
display(hr_employes)
```

```python
df_full["Join_date"]=pd.to_datetime(df_full["Join_Date"])
df_full["Join_year"]=df_full["Join_date"].dt.year
df_full["Join_month"]=df_full["Join_date"].dt.month
df_full["Join_day"]= df_full["Join_date"].dt.day
display(df_full[["Name", "Join_date","Join_year","Join_month","Join_day"]])
```

```python
df_sorted =df_full.sort_values("Join_date")
df_sorted[["Emp_ID","Salary","Join_date"]]
```

```python
df_sorted["Rolling_Avg_salary"]=df_sorted["Salary"].rolling(window=2).mean()
print("Rolling Average Salary:")
display(df_sorted[["Join_date","Salary","Rolling_Avg_salary"]])
```
