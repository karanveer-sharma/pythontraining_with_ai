```python
import numpy as np
import pandas as pd
```

```python
try:
  Titanic_data=pd.read_csv("Titanic-Dataset.csv")
  print("suceesfully upload")

except:
  print("Error")
```

```python
df_titanic = pd.DataFrame(Titanic_data)
display(df_titanic.head())
display(df_titanic.info())
```

```python
display(df_titanic.describe(include="all"))
```

```python
#missing age number
null_age = df_titanic["Age"].isnull().sum
print(f"missing values in age : {null_age()}")
```

```python
#total passengers survived
survive=df_titanic["Survived"].sum()
print(f"the number of survived : {survive}")
```

```python
# filling null values
df_titanic["Age"]=df_titanic["Age"].fillna(df_titanic["Age"].mean())
df_titanic
```

```python
df_titanic=df_titanic.drop("Cabin", axis = 1)
display(df_titanic.head())
```

```python
#percentage of female survivers compared to males
female_survive_total = df_titanic[df_titanic['Sex'] == 'female']['Survived'].sum()
female_percentatge = (female_survive_total / len(df_titanic[df_titanic['Sex'] == 'female'])) * 100
print(f"The total number of female survivors is = {female_percentatge:.2f}")

male_survive_total = df_titanic[df_titanic['Sex'] == 'male']['Survived'].sum()
male_percentatge = (male_survive_total / len(df_titanic[df_titanic['Sex'] == 'male'])) * 100
print(f"The total number of male survivors is = {male_percentatge}")

total_survivers = female_survive_total + male_survive_total
total_percentage = female_survive_total/ total_survivers *100
print(f"The female survivors in comparison to male survivors is = {total_percentage:.2f}%")
```

```python
#highest survivor class
class_survivors = df_titanic[df_titanic['Survived'] == 1]['Pclass'].value_counts()
highest_survivor_class = class_survivors.idxmax()
print(f"The class that has highest survivores = {highest_survivor_class} Class")
```

```python
#number of passanger which were alone
df_titanic["FamilySize"]= df_titanic["SibSp"]+df_titanic["Parch"]
alone = df_titanic[df_titanic["FamilySize"] ==0].shape[0]
print(f"the number of travelers which were alone is = {alone}")
```

```python
#Average fare paid by class 1
Average_fare = df_titanic[df_titanic["Pclass"]==1]["Fare"].mean()
print(f"Average fair paid by class 1 : {Average_fare}")
```

```python
#number of people under 18
children= df_titanic[df_titanic["Age"]<18].shape[0]
print(f"the number of passanger under 18 ={children}")
```

```python
#port embarked most passanger board the ship
port = df_titanic["Embarked"].value_counts().idxmax()
print(f"the port had most passangers : {port} port")
```

```python
#most common tittle
df_titanic["Title"] = df_titanic["Name"].str.extract(r",\s*([^\.]+)\.")
most_common_title = df_titanic['Title'].value_counts().idxmax()
print(f"the most common title = {most_common_title}")
```

```python
couples = df_titanic[(df_titanic['FamilySize'] == 1) & (df_titanic['Survived'] == 1)].shape[0]
print(f"The number of couples that survived is = {couples}")
```

```python
summary = {
    "questions" : [
        "Missing values in Age column",
        "Total survivors",
        "Percentage of female survivors",
        "Percentage of male survivors",
        "Female survivors in comparison to male survivors",
        "Class with highest survivors",
        "Passengers traveling alone",
        "Average fare in Pclass 1",
        "Children under 18",
        "Port with most passengers boarded",
        "Most common title",
        "Couples survived"
    ],
    "awnsers" : [
        null_age,
        survive,
        female_percentatge,
        male_percentatge,
        total_percentage,
        highest_survivor_class,
        alone,
        Average_fare,
        children,
        port,
        most_common_title,
        couples
    ]
}
summary_Data = pd.DataFrame(summary)
display(summary_Data)
```
